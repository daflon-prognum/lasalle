# Análise e Transformação de Dados com NumPy

```python
import numpy as np
```

# Atividade 1 — Análise de Dados no Varejo

```python
pecas_vendidas = np.array([150, 120, 90, 210, 300, 250, 180])
faturamento_diario = pecas_vendidas * 50
mascara_pico = pecas_vendidas > 200
dias_pico = pecas_vendidas[mascara_pico]
print(pecas_vendidas)
print(faturamento_diario)
print(mascara_pico)
print(dias_pico)
```

```python
vendas_filiais = np.array([
    [200, 220, np.nan, 250],
    [150, 180, 160, 190],
    [300, 310, 290, 330]
], dtype=float)
print(np.nansum(vendas_filiais, axis=1))
print(np.nanmean(vendas_filiais))
situacao = np.where(vendas_filiais >= 200, 'Meta Atingida', 'Abaixo')
situacao = np.where(np.isnan(vendas_filiais), 'Sem dado', situacao)
print(situacao)
```

```python
rng = np.random.default_rng(seed=42)
vendas_clientes = rng.choice(['Eletrônicos', 'Roupas', 'Casa'], size=50)
categorias, contagens = np.unique(vendas_clientes, return_counts=True)
print(dict(zip(categorias, contagens)))

faturamento_campanhas = np.array([12000, 45000, 23000, 89000, 31000])
indice = np.argmax(faturamento_campanhas)
print('Índice:', indice, '| Campanha:', indice + 1)
```

# Atividade 2 — Organização e Transformação de Dados

```python
dias_mes = np.arange(1, 31)
metas = np.linspace(20000, 30000, 5)
estoque = np.array([10.5, 20.1, 30.9]).astype(int)
ultimos = dias_mes[-5:][::-1]
print(dias_mes)
print(metas)
print(estoque)
print(ultimos)
```

```python
visitas = np.array([12000,13500,12800,14200,15000,15800,16500,17200,16900,18000,19500,21000])
trimestres = visitas.reshape(4, 3)
primeiro_semestre = np.array([[120,135,128],[142,150,158]])
segundo_semestre = np.array([[165,172,169],[180,195,210]])
ano_completo = np.vstack([primeiro_semestre, segundo_semestre])
primeiro_trimestre = trimestres[0].copy()
relatorio = trimestres.ravel()
print(trimestres)
print(ano_completo)
print(primeiro_trimestre)
print(relatorio)
```

```python
pontuacoes = np.array([85, 92, 78, 95, 88])
ranking = np.argsort(pontuacoes)[::-1]
print(ranking)
print(pontuacoes[ranking])

precos_base = np.array([100, 200, 300])
precos_coluna = precos_base[:, np.newaxis]
descontos = np.array([0.0, 0.10, 0.20])
print(precos_coluna * (1 - descontos))

A = np.array([[2, 1], [1, -1]], dtype=float)
b = np.array([500, 100], dtype=float)
print(np.linalg.solve(A, b))
```
